```python
#Import libraries
import pandas as pd
import numpy as np
```

```python
#Load Dataset
df = pd.read_excel("Ecommerce_Unclean_Project.xlsx")
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
#Replace N/A and NULL with NaN
df.replace(["N/A","NULL"," "],np.nan,inplace=True)
```

```python
#Remove Extra Spaces
df = df.apply(lambda x: x.str.strip() if x.dtype =='str' else x)
```

```python
#Proper Case
df["Customer_Name"]=df["Customer_Name"].str.title()

df["City"]=df["City"].str.title()

df["State"]=df["State"].str.title()
```

```python
#Fix Emails
df=df[df["Email"].str.contains("@",na=False)]
```

```python
#Convert Date Columns
df["Order_Date"]=pd.to_datetime(df["Order_Date"],errors="coerce")

df["Delivery_Date"]=pd.to_datetime(df["Delivery_Date"],errors="coerce")
```

```python
#Convert Numeric Columns
cols = ["Qty","Unit_Price","Discount"]

for c in cols:
    df[c]=pd.to_numeric(df[c],errors="coerce")
```

```python
#Remove Negative Quantity
df=df[df["Qty"]>0]
```

```python
#Fill Missing Values
df["Discount"] = df["Discount"].fillna(0)

df["Phone"] = df["Phone"].fillna("Unknown")

df["Delivery_Date"] = df["Delivery_Date"].fillna(df["Order_Date"])
```

```python
#Feature Engineering
# Total sales
df["Sales"]=df["Qty"]*df["Unit_Price"]
```

```python
df.head()
```

```python
#Net Amount
df["Net_Amount"]=df["Sales"]-(df["Sales"]*df["Discount"]/100)
```

```python
df.head()
```

```python
#Profit (Example; 20%)
df["Profit"]=df["Net_Amount"]*0.20
```

```python
df.head()
```

```python
#Order Month
df["Month"]=df["Order_Date"].dt.month_name()
```

```python
df.head()
```

```python
# Order Year
df["Year"]=df["Order_Date"].dt.year
```

```python
df.head()
```

```python
#Weekday
df["Weekday"]=df["Order_Date"].dt.day_name()
```

```python
#Export Clean Data
df.to_csv("Clean_Ecommerce.csv",index=False)
```
